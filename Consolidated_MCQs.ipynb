# 11 Horizontal and vertical scaling

## Q1

Consider an application that has worldwide users and servers in only one location. As the traffic increases which of the following factors should the application be scaled for?

1. Storage
2. Throughput
3. Parallelism
4. Geolocation


<details>
<summary>A1</summary>

Answer: 4. Geolocation

</details>
